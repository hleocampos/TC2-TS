### Chebyshev

<img src="./Simulacion_Cheby.PNG" width="500" /> 
(Simulación)
<img src="./Cheby.PNG" width="500" /> 
(Practica)


```python
from PIL import Image
from IPython.display import display

img1 = Image.open("Simulacion_comparacion_Cheby_v1.PNG").convert("RGBA")
img2 = Image.open("./Cheby.PNG").convert("RGBA")

img2 = img2.resize((220, 300))


x = 210
y = 150

img2.putalpha(
    img2.getchannel("A").point(lambda p: int(p * 0.8))
)

img1.alpha_composite(img2, (x, y))


display(img1)
```

Como se puede ver, lo obtenido en la practica es casi igual a lo simulado. Para más precision se tuviesen que haber medido más puntos. Pero el osciloscopio no medía debajo de 40dB con la tension aplicada. (fs=2khz)

### Notch

<img src="./Simulacion_Notch.PNG" width="500" /> 
(Simulacion)
<img src="./Notch.PNG" width="500" /> 
(Practica)

```python
from PIL import Image
from IPython.display import display


img1 = Image.open("./Simulacion_Notch.PNG").convert("RGBA")
img2 = Image.open("./Notch.PNG").convert("RGBA")


img2 = img2.resize((1670, 100))


x = 30
y = 60

img2.putalpha(
    img2.getchannel("A").point(lambda p: int(p * 0.8))
)

img1.alpha_composite(img2, (x, y))


display(img1)
```

El filtro notch en la practica no llega al mismo nivel de atenuacion que el simulado, por temas de medición. Pero el ancho de banda de 1hz y frecuencia central en 50hz se respetan. (fs =200hz)
