```python
from pytc2.sistemas_lineales import analyze_sys, bodePlot, pzmap, pretty_print_bicuad_omegayq, pretty_print_SOS, pretty_print_lti
from pytc2.general import print_latex, a_equal_b_latex_s
```

```python
# Comienzo de la simulación

w0 = 1

# Cargamos la funcion transferencia como vectores de sus coeficientes.

num = np.array([ 1, -w0 ])
den = np.array([ 1, w0])

H1 = sig.TransferFunction( num, den )

# mostramos la transferencia construida
display(H1)
```

```python
num = [ 1, -w0 ]
den = [ 1, w0]
pretty_print_lti(num,den)

# o un poco mejor, presentarlo como una ecuación:
print_latex(a_equal_b_latex_s('H_d(s)', pretty_print_lti(num,den,displaystr=False)))
```
